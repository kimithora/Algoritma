Inheritance

```python
class parent:
    def hai (self):
        print("This is the parent class")

class child(parent):
    pass

obj = child()
obj.hai()
```

```python
class parent:
    def greet (self):
        print("Good morning")
    
class child(parent):
    def greet (self):
        print("Good afternoon")

c = child()
c.greet()
```

```python
class parent:
    def greetparent (self):
        print("Good morning")
    
class child(parent):
    def greet (self):
        print("Good afternoon")

c = child()
c.greet()
c.greetparent()
```

```python
class A:
    def methode_a(self):
        print("methode a")

class B:
    def methode_b(self):
        print("methode b")

class C(A,B):
    pass

obj = C()
obj.methode_a()
obj.methode_b()
```

```python
class employe:
    def __init__ (self, name):
        self.name = name

    def work(self):
        print(f"{self.name} sedang bekerja")

class dataanalyst(employe):
    def work (self):
        print (f"{self.name} lagi ngitung")

class manager(employe):
    def work (self):
        print (f"{self.name} lagi tidur")

A = dataanalyst("kimi")
B = manager("thora")

A.work()
B.work()
```

Polymorphism

```python
class Employe:
    def __init__(self, salary):
        self.salary = salary

    def calc(self):
        return self.salary


class DataAnalyst(Employe):
    def calc(self):
        bonus = self.salary * 0.15
        return self.salary + bonus


class Manager(Employe):
    def calc(self):
        bonus = self.salary * 0.25
        return self.salary + bonus


def main():
    employees = [
        DataAnalyst(5000000),
        Manager(10000000)
    ]

    for x in employees:
        print(f"role: {x.__class__.__name__}")
        print(f"salary: {x.calc()}")


if __name__ == "__main__":
    main()
```

```python
class shape:
    def area(self):
        pass

class circle(shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 22/7*(self.radius**2)

class square(shape):
    def __init__(self, side):
        self.side = side 

    def area(self):
        return self.side**2
    
def main():
    lingkaran = circle(7)
    persegi = square(13)

    print (f"luas lingkaran adalah {lingkaran.area()}")
    print (f"luas persegi adalah {persegi.area()}")

if __name__ == "__main__":
    main()
```
