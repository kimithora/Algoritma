# Abstraction

```python
from abc import ABC, abstractmethod

class Kendaraan(ABC):
    
    @abstractmethod
    def jalan(self):
        pass
    
    @abstractmethod
    def bahan_bakar(self):
        pass

class Mobil(Kendaraan):
    
    def jalan(self):
        return "Mobil berjalan dengan roda empat"
    
    def bahan_bakar(self):
        return "Mobil menggunakan bensin"

class SepedaMotor(Kendaraan):
    
    def jalan(self):
        return "Motor berjalan dengan dua roda"
    
    def bahan_bakar(self):
        return "Motor menggunakan bensin"

# Implementasi
kendaraan1 = Mobil()
kendaraan2 = SepedaMotor()

print(kendaraan1.jalan())
print(kendaraan2.bahan_bakar())
```

```python
from abc import ABC, abstractmethod

class AkunBank(ABC):    
    def __init__(self, nama, saldo):
        self.nama = nama
        self.saldo = saldo
    
    @abstractmethod
    def hitung_bunga(self):
        pass
    
    @abstractmethod
    def info_akun(self):
        pass

class Tabungan(AkunBank):
    def hitung_bunga(self):
        return self.saldo * 0.02  
    
    def info_akun(self):
        return f"Akun Tabungan - {self.nama}, Saldo: {self.saldo}"

class Deposito(AkunBank):
    def hitung_bunga(self):
        return self.saldo * 0.05  
    
    def info_akun(self):
        return f"Akun Deposito - {self.nama}, Saldo: {self.saldo}"
```

```python
akun1 = Tabungan("Kimi", 1000000)
akun2 = Deposito("Kimi", 1000000)

print(akun1.info_akun())
print("Bunga:", akun1.hitung_bunga())

print(akun2.info_akun())
print("Bunga:", akun2.hitung_bunga())
```

```python
from abc import ABC, abstractmethod

class Bank(ABC):
    def __init__(self, nama, nominal, bunga):
        self.nama = nama
        self.nominal = nominal
        self.bunga = bunga

    @abstractmethod
    def proses(self):
        pass


class Tabungan(Bank):
    def proses(self):
        uang = self.nominal + (self.nominal*self.bunga)
        return f"Tabungan {self.nama} sejumlah {uang} dengan bunga {self.bunga}"


class Pinjaman(Bank):
    def proses(self):
        minjam = self.nominal + (self.nominal*self.bunga)
        return f"Pinjaman oleh {self.nama} dengan total {minjam}"


class Reksadana(Bank):
    def proses(self):
        saham = self.nominal + (self.nominal*self.bunga)
        return f"Reksadana {self.nama} sebesar {saham} dengan bunga {self.bunga}"


akun1 = Pinjaman("Kimi", 123000000, 0.13)
akun2 = Tabungan("James", 45000, 0.3)

print(akun1.proses())
print(akun2.proses())
```

```python
from abc import ABC, abstractmethod

class SektorEkonomi(ABC):
    def __init__ (self, nama_sektor, kontribusi, investasi):
        self.nama_sektor = nama_sektor
        self.kontribusi = kontribusi
        self.investasi = investasi

    @abstractmethod
    def hitung_pertumbuhan(self):
        pass

    @abstractmethod
    def kebijakan(self):
        pass

class SektorPariwisata(SektorEkonomi):
    def hitung_pertumbuhan(self):
        hasil = self.investasi * 2
        return hasil
    
    def kebijakan(self):
        return "pemerintah meningkatkan promosi wisata dan pembangunan infrastruktur"

pariwisata = SektorPariwisata(
    "pariwisata", 
    "sumber devisa negara", 
    200
)

print("=== Sektor Pariwisata ====")
print(f"jenis sektor: {pariwisata.nama_sektor}")
print(f"kontribusi: {pariwisata.kontribusi}")
print(f"investasi: {pariwisata.investasi}")

print("\nHasil Perhitungan")
print(f"pertumbuhan sektor wisata sebesar {pariwisata.hitung_pertumbuhan()} milyar rupiah")

print("\nKebijakan Pemerintah:")
print(pariwisata.kebijakan())
```

```python
from abc import ABC, abstractmethod

class PelayananRumahsakit(ABC):
    def __init__ (self,nama_pasien, jenis_pelayanan, biaya_dasar, jumlah_hari, jumlah_kunjungan):
        self.nama_pasien = nama_pasien
        self.jenis_pelayanan = jenis_pelayanan
        self.biaya_dasar = biaya_dasar
        self.jumlah_hari = jumlah_hari
        self.jumlah_kunjungan = jumlah_kunjungan
    @abstractmethod
    def hitung_biaya(self):
        pass

    @abstractmethod
    def keterangan(self):
        pass

class RawatInap(PelayananRumahsakit):
    def hitung_biaya(self):
        total = self.biaya_dasar + (self.jumlah_hari * 5000000)
        return total
    
    def keterangan(self):
        return "pasien sedang rawat inap"
    
class RawatJalan(PelayananRumahsakit):
    def hitung_biaya(self):
        total = self.biaya_dasar + (self.jumlah_kunjungan * 150000)
        return total
    
    def keterangan(self):
        return "pasien sedang rawat jalan" 
    
class Operasi(PelayananRumahsakit):
    def hitung_biaya(self):
        total = self.biaya_dasar + 50000000
        return total
    
    def keterangan(self):
        return "pasien sedang operasi berat"

pasien1 = Operasi(
    "Andi",
    "Operasi",
    "3000000",
    13,
    3
)

print("=== Pasien Operasi ===")
print(f"nama pasien: {pasien1.nama_pasien}")
print(f"Jenis Pelayanan: {pasien1.jenis_pelayanan}")
print(f"biaya dasar: {pasien1.biaya_dasar}")

print("\nTotal Biaya:")
print(f"Total biaya operasi sebesar {Operasi.hitung_biaya}")

print("\nKeterangan:")
print(pasien1.keterangan())
```

# Encapsulation

```python
class Mahasiswa:
    def __init__ (self, nama, nim, nilai_tugas, nilai_uts, nilai_uas):
        self.nama = nama
        self.nim = nim
        self.__nilai_tugas = nilai_tugas
        self.__nilai_uts = nilai_uts
        self.__nilai_uas = nilai_uas

    def set_nilaitugas(self,nilai_tugas):
        self.__nilai_tugas = nilai_tugas

    def set_nilaiuts(self,nilai_uts):
        self.__nilai_uts = nilai_uts

    def set_nilaiuas(self, nilai_uas):
        self.__nilai_uas = nilai_uas
    
    def get_nilaitugas(self):
        return self.__nilai_tugas
    
    def get_nilaiuts(self):
        return self.__nilai_uts
    
    def get_nilaiuas(self):
        return self.__nilai_uas
    
    def hitung_nilai_akhir(self):
        nilai_akhir = (self.__nilai_tugas * 0.30) + (self.__nilai_uts * 0.30) + (self.__nilai_uas * 0.40)
        return nilai_akhir
    
    def status_kelulusan (self):
        if self.hitung_nilai_akhir() >= 90:
            return "A dan Lulus"
        elif self.hitung_nilai_akhir() >= 70:
            return "B dan Lulus"
        elif self.hitung_nilai_akhir() >= 60:
            return "C dan Mengulang ujian"
        elif self.hitung_nilai_akhir() >= 40:
            return "D dan tidak lulus"
        else:
            return "mengulang semester depan"
        
mahasiswa = Mahasiswa(
    "Budi",
    "243842",
    80,
    90,
    65
)

print(mahasiswa.nama)
print(mahasiswa.nim)
print(mahasiswa.get_nilaitugas())
print(mahasiswa.get_nilaiuts())
print(mahasiswa.get_nilaiuas())

print(mahasiswa.hitung_nilai_akhir())
print(mahasiswa.status_kelulusan())
```

```python
class RekeningBank:
    def __init__ (self,nama, nomor_rekening, saldo):
        self.nama = nama
        self._nomor_rekening = nomor_rekening
        self.__saldo = saldo

    def set_saldo(self,saldo):
        self.__saldo = saldo

    def get_saldo(self):
        return self.__saldo
    
    def setor_uang(self,jumlah):
        self.__saldo += jumlah
        return f"setor uang sejumlah {self.__saldo}"
    
    def tarik_uang(self, jumlah):
        if jumlah <= self.__saldo:
            self.__saldo -= jumlah
            return f"tarik uang sebesar {self.__saldo} berhasil"
        

nasabah1 = RekeningBank(
    "Udin",
    "123584773", 
    134000000
)

print(f"nama nasabah : {nasabah1.nama}")
print(f"No. Rekening : {nasabah1._nomor_rekening}")
print(f"Saldo Awal : {nasabah1.get_saldo()}")

print(f"\nSetor uang : {nasabah1.setor_uang(13000000)}")
print(f"\nTarik Uang: {nasabah1.tarik_uang(300000)}")


```
