# Abstraction

```python
from abc import ABC, abstractmethod

class Kendaraan(ABC):
    
    @abstractmethod
    def jalan(self):
        pass
    
    @abstractmethod
    def bahan_bakar(self):
        pass

class Mobil(Kendaraan):
    
    def jalan(self):
        return "Mobil berjalan dengan roda empat"
    
    def bahan_bakar(self):
        return "Mobil menggunakan bensin"

class SepedaMotor(Kendaraan):
    
    def jalan(self):
        return "Motor berjalan dengan dua roda"
    
    def bahan_bakar(self):
        return "Motor menggunakan bensin"

# Implementasi
kendaraan1 = Mobil()
kendaraan2 = SepedaMotor()

print(kendaraan1.jalan())
print(kendaraan2.bahan_bakar())
```

```python
from abc import ABC, abstractmethod

class AkunBank(ABC):    
    def __init__(self, nama, saldo):
        self.nama = nama
        self.saldo = saldo
    
    @abstractmethod
    def hitung_bunga(self):
        pass
    
    @abstractmethod
    def info_akun(self):
        pass

class Tabungan(AkunBank):
    def hitung_bunga(self):
        return self.saldo * 0.02  
    
    def info_akun(self):
        return f"Akun Tabungan - {self.nama}, Saldo: {self.saldo}"

class Deposito(AkunBank):
    def hitung_bunga(self):
        return self.saldo * 0.05  
    
    def info_akun(self):
        return f"Akun Deposito - {self.nama}, Saldo: {self.saldo}"
```

```python
akun1 = Tabungan("Kimi", 1000000)
akun2 = Deposito("Kimi", 1000000)

print(akun1.info_akun())
print("Bunga:", akun1.hitung_bunga())

print(akun2.info_akun())
print("Bunga:", akun2.hitung_bunga())
```

```python
from abc import ABC, abstractmethod

class Bank(ABC):
    def __init__(self, nama, nominal, bunga):
        self.nama = nama
        self.nominal = nominal
        self.bunga = bunga

    @abstractmethod
    def proses(self):
        pass


class Tabungan(Bank):
    def proses(self):
        uang = self.nominal + (self.nominal*self.bunga)
        return f"Tabungan {self.nama} sejumlah {uang} dengan bunga {self.bunga}"


class Pinjaman(Bank):
    def proses(self):
        minjam = self.nominal + (self.nominal*self.bunga)
        return f"Pinjaman oleh {self.nama} dengan total {minjam}"


class Reksadana(Bank):
    def proses(self):
        saham = self.nominal + (self.nominal*self.bunga)
        return f"Reksadana {self.nama} sebesar {saham} dengan bunga {self.bunga}"


akun1 = Pinjaman("Kimi", 123000000, 0.13)
akun2 = Tabungan("James", 45000, 0.3)

print(akun1.proses())
print(akun2.proses())
```

```python
from abc import ABC, abstractmethod

class SektorEkonomi(ABC):
    def __init__ (self, nama_sektor, kontribusi, investasi):
        self.nama_sektor = nama_sektor
        self.kontribusi = kontribusi
        self.investasi = investasi

    @abstractmethod
    def hitung_pertumbuhan(self):
        pass

    @abstractmethod
    def kebijakan(self):
        pass

class SektorPariwisata(SektorEkonomi):
    def hitung_pertumbuhan(self):
        hasil = self.investasi * 2
        return hasil
    
    def kebijakan(self):
        return "pemerintah meningkatkan promosi wisata dan pembangunan infrastruktur"

pariwisata = SektorPariwisata(
    "pariwisata", 
    "sumber devisa negara", 
    200
)

print("=== Sektor Pariwisata ====")
print(f"jenis sektor: {pariwisata.nama_sektor}")
print(f"kontribusi: {pariwisata.kontribusi}")
print(f"investasi: {pariwisata.investasi}")

print("\nHasil Perhitungan")
print(f"pertumbuhan sektor wisata sebesar {pariwisata.hitung_pertumbuhan()} milyar rupiah")

print("\nKebijakan Pemerintah:")
print(pariwisata.kebijakan())
```
